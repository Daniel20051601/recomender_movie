```python
# ratings unicos
df_titles['rating'] = df_titles['rating'].astype('category')

display(df_titles['rating'].value_counts())
```

```python
display(df_titles[(df_titles['rating'].str.contains('min', na = False, case = False))])
```

```python
# First we need to move 'min' values from rating into the duration column
df_titles.loc[df_titles['rating'].str.contains('min', na = False), 'duration'] = df_titles['rating']

# Then, set the rating for 'TV-MA'(in this case)
df_titles.loc[df_titles['rating'].str.contains('min', na = False), 'rating'] = 'TV-MA'

# Here are the updated rows
display(df_titles[(df_titles['director'].str.contains('Louis C.K.', na = False))])
```

```python
print('Pelicula o serie mas antigua')
display(df_titles[(df_titles['release_year'] == min(df_titles['release_year']))])

print('Pelicula o serie mas reciente')
display(df_titles[(df_titles['release_year'] == max(df_titles['release_year']))])
```

```python
# This function validate the date
from datetime import datetime

def check_format_date(date_str):
    try:
        datetime.strptime(date_str.strip(), '%B %d, %Y')
        return True
    except (ValueError, TypeError, AttributeError):
        return False
```

```python
# Top 10 countries with more content
df_top_countries = df_titles['country'].explode().value_counts().head(10).reset_index(name='total')
df_top_countries.columns = ['country','total']
display(df_top_countries)
```

```python
df_titles.info()
```

```python
# Cantidad de contenido por año
added_by_year = df_titles['date_added'].dt.year.value_counts().reset_index()
added_by_year.columns = ['Year', 'Total']
display(added_by_year.sort_values('Year'))
```

```python
# Genero mas comun
genero_mas_comun = df_titles['listed_in'].explode().value_counts().head(1).reset_index()
genero_mas_comun.columns = ['Genero', 'Cantidad']
display(genero_mas_comun)
```

```python
#Director con mas titulos
director_mas_repetido = df_titles['director'].str.split(", ").explode().value_counts().idxmax()
content_by_director = df_titles[df_titles['director'].fillna('').str.contains(director_mas_repetido)]
display(content_by_director[['title','director','release_year','description']])
```

```python
#Duracion promedio de una pelicula
df_movies = df_titles[df_titles['type'] == 'Movie'].copy()
df_movies['duration_num'] = df_movies['duration'].str.replace('min','').astype('float')
tiempo_promedio = df_movies['duration_num'].mean()

print(f"Duracion promedio de las peliculas: {tiempo_promedio:.2f} minutos")
```
