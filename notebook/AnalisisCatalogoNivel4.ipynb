```python
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os
```

```python
df_titles = pd.read_csv(r"D:\Data Engineer\NetflixTitlesNivel4.csv")
df_titles.head()
```

```python
# Which country produces more movies vs series?
df_countries = df_titles[['type','country']].copy()
df_countries = df_countries.explode('country')

tabla = df_countries.groupby(['country','type']).size().unstack(fill_value=0)

pais_mas_peliculas = tabla['Movie'].idxmax()
pais_mas_series = tabla['TV Show'].idxmax()

print('Pais con mas peliculas: ', pais_mas_peliculas, "-", int(tabla.loc[pais_mas_peliculas, 'Movie']))
print('Pais con mas series: ', pais_mas_series, "-", int(tabla.loc[pais_mas_series, 'TV Show' ]))
```

```python
df_movies = df_titles[(df_titles['type'] == 'Movie')]
df_movies.shape
```

```python
load_dotenv()
#Connect to the database
engine = create_engine(os.getenv("DATABASE_URL"))
```

```python
#Tested the connection
conn = engine.connect()
```

```python
display(df_movies.isnull().sum())
```

```python
#Fill empty fields with Unknown
for column in ['director','cast','country','rating']:
    if column in df_movies.columns:
        df_movies[column] = df_movies[column].fillna('Unkown')

display(df_movies.isnull().sum())
```

```python
#Added the table to the database
df_movies.to_sql('Movies', engine, if_exists='replace',index=False)
```
