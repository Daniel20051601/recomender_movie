```python
import pandas as pd
import ast
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
#Connected to the database
engine = create_engine(os.getenv("DATABASE_URL"))
#Tested the connection
engine.connect()
```

```python
#fetched the movies table
df_content = pd.read_sql('SELECT * FROM "Movies"', engine)
df_content['listed_in']
```

```python
df_content['listed_in'] = df_content['listed_in'].apply(ast.literal_eval)

categorias = df_content['listed_in'].explode().drop_duplicates().reset_index(drop=True)

df_categorias = categorias.to_frame(name='categoria')

df_categorias
```

```python
# Created new table `categoria`
df_categorias.to_sql('categoria', engine, if_exists='replace', index=False)
```
